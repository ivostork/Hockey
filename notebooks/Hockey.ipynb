## Cíl

**Úvod**  
Vžijeme se do role analytika v nově vzniklé online sázkové kanceláři. Naším úkolem bude získat výsledky hokejových lig z veřejně dostupných zdrojů a následně určit počáteční sázkové kurzy, které budou aktualizovány s přicházejícími novými daty.

Abychom strukturovali práci na projektu, rozdělili jsme ji do následujících čtyř částí tohoto notebooku:

- **Stažení dat** – v tomto kroku si stáhneme data ze stránky Scrape This Site (část 1),

- **Zpracování dat** – budeme analyzovat HTML stránky a připravíme data pro analýzu (část 2),

- **Analýza dat** – provedeme explorační analýzu dat, abychom se seznámili s jejich charakteristikami (část 3),

- **Doporučení** – vytvoříme počáteční sázkové kurzy a prodiskutujeme možné způsoby dalšího zlepšení našeho postupu (část 4).
